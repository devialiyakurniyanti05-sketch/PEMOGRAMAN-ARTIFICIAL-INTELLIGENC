```python
tahun = int(input("tahun: "))

if (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0):
    print(f"tahun {tahun}, merupakan tahun kabisat")
else:
    print(f"tahun {tahun}, bukan tahun kabisat")
```

```python
tanggal = int(input("tanggal anda lahir: "))
bulan = input("bulan anda lahir: ").lower()

if bulan == "januari":
    zodiak = "Capricorn" if tanggal <= 19 else "Aquarius"
elif bulan == "februari":
    zodiak = "Aquarius" if tanggal <= 18 else "Pisces"
elif bulan == "maret":
    zodiak = "Pisces" if tanggal <= 20 else "Aries"
elif bulan == "april":
    zodiak = "Aries" if tanggal <= 19 else "Taurus"
elif bulan == "mei":
    zodiak = "Taurus" if tanggal <= 20 else "Gemini"
elif bulan == "juni":
    zodiak = "Gemini" if tanggal <= 20 else "Cancer"
elif bulan == "juli":
    zodiak = "Cancer" if tanggal <= 22 else "Leo"
elif bulan == "agustus":
    zodiak = "Leo" if tanggal <= 22 else "Virgo"
elif bulan == "september":
    zodiak = "Virgo" if tanggal <= 22 else "Libra"
elif bulan == "oktober":
    zodiak = "Libra" if tanggal <= 22 else "Scorpio"
elif bulan == "november":
    zodiak = "Scorpio" if tanggal <= 21 else "Sagittarius"
elif bulan == "desember":
    zodiak = "Sagittarius" if tanggal <= 21 else "Capricorn"
else:
    zodiak = None

if zodiak:
    print(f"Astrologi anda adalah {zodiak}")
else:
    print("Bulan tidak valid")
```
