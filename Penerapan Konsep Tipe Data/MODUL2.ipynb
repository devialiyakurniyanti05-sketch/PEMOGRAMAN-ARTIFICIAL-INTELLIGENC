```python


# Praktikum Pemrograman Artificial Intelligence
# Variabel dan Tipe Data

print("Praktikum Pemrograman Artificial Intelligence")
print("Nama: Devi Aliya Kurniyanti")
print("Umur:", 21)

# Tipe data String
nama = "Devi Aliya Kurniyanti"
print(nama)
print(type(nama))

# Tipe data Integer
umur = 21
print(umur)
print(type(umur))

# Tipe data Boolean
mahasiswa = True
print(mahasiswa)
print(type(mahasiswa))

# Tipe data Float
nilai = 90.5
print(nilai)
print(type(nilai))

# Tipe data List
hobi = ["berjualan", "mendengarkan musik"]
print(hobi)

# Tipe data Dictionary
biodata = {
    "nama": "Devi Aliya Kurniyanti",
    "umur": 21
}
print(biodata)
print(type(biodata))

```
